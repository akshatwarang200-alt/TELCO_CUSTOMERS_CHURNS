```python
CS = df.reset_index()                       # CustomerID becomes a normal column
print("Rows:", len(CS), "| duplicate IDs:", CS["CustomerID"].duplicated().sum())

# Look at the duplicate IDs: identical or conflicting?
print(CS[CS["CustomerID"].duplicated(keep=False)].sort_values("CustomerID").head(10))

CS = CS.drop_duplicates(subset="CustomerID", keep="first").reset_index(drop=True)
print("Rows after:", len(CS))            # should be about 10,000
```

```python
CS.isnull().sum()
```

```python
CS
```

```python
print((CS["MonthlyCharges"] < 0).sum())
CS.loc[CS["MonthlyCharges"] < 0, "MonthlyCharges"] = np.nan
upper = CS["MonthlyCharges"].quantile(0.99)
CS["MonthlyCharges"] = CS["MonthlyCharges"].clip(upper=upper)
CS["MonthlyCharges"] = CS["MonthlyCharges"].fillna(CS["MonthlyCharges"].median())
```

```python
CS["SignupDate"] = pd.to_datetime(CS["SignupDate"])
CS["TenureMonths"] = (pd.Timestamp("2023-12-31") - CS["SignupDate"]).dt.days // 30
CS
```

```python
usage = (df1.reset_index().groupby("CustomerID").agg(
            AvgCallMinutes=("CallMinutes", "mean"),
            AvgDataGB=("DataUsageGB", "mean"),
            AvgSMS=("SMSCount", "mean"),
            TotalComplaints=("Complaints", "sum"),
         ).reset_index())
usage
```

```python
labels = df2.reset_index()
final = (CS.merge(usage, on="CustomerID", how="inner")
           .merge(labels, on="CustomerID", how="inner"))
print(final.shape, final["Churn"].mean())
```

```python
final
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(4.5, 4))
ax = sns.countplot(data=final, x="Churn")
for p in ax.patches:
    ax.annotate(f"{p.get_height() / len(final):.1%}",
                (p.get_x() + p.get_width() / 2, p.get_height()), ha="center", va="bottom")
plt.title("Churn distribution (0 = stayed, 1 = left)")
plt.show()
```

inference : About 7.9% of customers churned, so the data is imbalanced and accuracy alone would be misleading.

```python
num_cols = ["Age", "MonthlyCharges", "TenureMonths",
            "AvgCallMinutes", "AvgDataGB", "AvgSMS", "TotalComplaints"]
cat_cols = ["Gender", "Region", "ContractType"]
```

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, c in zip(axes.flat, num_cols):
    sns.histplot(final[c], kde=True, ax=ax)
    ax.set_title(f"Distribution of {c}")
axes.flat[-1].axis("off")
plt.tight_layout(); plt.show()
```

inference : MonthlyCharges is skewed with extreme values up to 390, which I capped at 99. Age is spread between 18 to 69.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, c in zip(axes, cat_cols):
    rate = final.groupby(c)["Churn"].mean().sort_values(ascending=False) * 100
    sns.barplot(x=rate.index, y=rate.values, ax=ax)
    ax.set_title(f"Churn rate (%) by {c}"); ax.set_ylabel("Churn %")
    for i, v in enumerate(rate.values):
        ax.text(i, v, f"{v:.1f}", ha="center", va="bottom")
plt.tight_layout(); plt.show()
```

inference : Yearly contracts churn at 8.8% compared with 7.6% for monthly contracts, so monthly contract type customers are the highest risk

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, c in zip(axes.flat, ["Age", "MonthlyCharges", "TenureMonths",
                             "AvgCallMinutes", "AvgDataGB", "TotalComplaints"]):
    sns.boxplot(data=final, x="Churn", y=c, ax=ax)
    ax.set_title(f"{c} by churn")
plt.tight_layout(); plt.show()
```

```python
final["AgeGroup"] = pd.cut(final["Age"], [17, 25, 35, 45, 55, 70],
                           labels=["18-25", "26-35", "36-45", "46-55", "56+"])
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=final, x="AgeGroup", y="Churn", ax=axes[0], errorbar=None)
axes[0].set_title("Churn rate by age group")
sns.barplot(data=final, x="TotalComplaints", y="Churn", ax=axes[1], errorbar=None)
axes[1].set_title("Churn rate by total complaints")
plt.tight_layout(); plt.show()
```

inference : Churn rate rises from 7% with 0 complaints to 11% to 12% with 13 to 14 complaints, so complaints are an early warning signal.

```python
pivot = final.pivot_table(index="ContractType", columns="Region",
                          values="Churn", aggfunc="mean") * 100
plt.figure(figsize=(6, 4))
sns.heatmap(pivot, annot=True, fmt=".1f", cmap="YlOrRd")
plt.title("Churn rate (%) by contract type and region")
plt.show()
```

inference : The riskiest segment is Yearly contract customers in Rural areas at 9.4% churn ,The safest segment is Prepaid customers in Rural areas at 6.7%.
